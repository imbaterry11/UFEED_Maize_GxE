# UFEED models yield prediction for Maize GxE competition

The models were trained using AutoGluon(1.1.0) in Python 3.10.14. However, the training data can generate a prediction model using the most updated AutoGluon package with any supported versions of Python.

```python
import autogluon
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
from autogluon.tabular import TabularDataset, TabularPredictor
from autogluon.core.metrics import make_scorer
```

## Mean yield per env model prediction
The goal of this step is to use a model named 'maize_yield_model_env' to predict mean yield per env. 

```python
predictor_maize_env = TabularPredictor.load("maize_yield_model_env",require_version_match=False)
```

```python
test_data_env = pd.read_csv('yield_data_predict.csv', sep=",", header=0) 
```

```python
y_pred_env = predictor_maize_env.predict(test_data_env)
```

```python
y_pred_env.to_csv (r'yield_data_env_pred.csv', index = True, header=True)
```

## Genotpye effect per env model prediction
The goal of this step is to use a model named 'maize_yield_model_geno' to predict genotpye effect per env.

```python
predictor_maize_geno = TabularPredictor.load("maize_yield_model_geno",require_version_match=False)
```
